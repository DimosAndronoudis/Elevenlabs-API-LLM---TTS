```python
ELEVENLABS_API_KEY = "sk_25d66eb2ea959900de5230ab5f3a0b869f18bd83b16c7736"


# Built-in Voice ID

voice_id = "21m00Tcm4TlvDq8ikWAM"  # Rachel


# Theme-Based Text Generator

generator = pipeline("text-generation", model="gpt2")
```

```python
def generate_sentences_about(theme, max_length=100):
    prompt = f"Write a few interesting sentences about {theme}."
    result = generator(prompt, max_length=max_length, num_return_sequences=1)[0]['generated_text']
    return result.strip()


theme = "time travel"
text_to_speak = generate_sentences_about(theme)
print("Text to synthesize:\n", text_to_speak)



```

```python

# Call ElevenLabs Text-to-Speech API


def synthesize_speech(voice_id, text, model="eleven_multilingual_v2"):
    url = f"https://api.elevenlabs.io/v1/text-to-speech/{voice_id}"
    headers = {
        "xi-api-key": ELEVENLABS_API_KEY,
        "Content-Type": "application/json"
    }
    payload = {
        "text": text,
        "model_id": model,
        "voice_settings": {
            "stability": 0.75,
            "similarity_boost": 0.75
        }
    }
    response = requests.post(url, json=payload, headers=headers)
    if response.status_code == 200:
        return response.content
    else:
        raise Exception(f"Failed to synthesize speech: {response.text}")

```

```python
# Run TTS and Save Audio
audio = synthesize_speech(voice_id, text_to_speak)
with open("output.wav", "wb") as f:
    f.write(audio)


# Playback Audio

print("\nGenerated Audio:")
display(Audio("output.wav"))
```
